```python
import torch
from hydra import compose, initialize_config_dir
from omegaconf import OmegaConf

from baseformer.main import initialize_data, initialize_training
from baseformer.tokenization.bpe import BPETokenizer
```

```python
# Load Hydra config programmatically
# Use absolute path to config directory
config_dir = os.path.abspath("baseformer/configs")

with initialize_config_dir(version_base=None, config_dir=config_dir):
    # You can add overrides as a list of strings, e.g.:
    # overrides=["model=medium", "experiment.batch_size=32"]
    cfg = compose(config_name="config", overrides=[])

# Print resolved config
print(OmegaConf.to_yaml(cfg))
```

```python
# Initialize model, optimizer, lr_schedule, and RoPE
model, optimizer, lr_schedule, rope = initialize_training(cfg)

num_params = sum(p.numel() for p in model.parameters())
print(f"Model parameters: {num_params:,}")
```

```python
# Initialize data loaders (optional - comment out if not needed)
train_loader, val_loader = initialize_data(cfg)
train_iter = iter(train_loader)


# Load tokenizer for sample generation
tokenizer = BPETokenizer.from_files(
    cfg.data.tokenizer.vocab_path,
    cfg.data.tokenizer.merges_path,
)
```

```python
batch = next(train_iter)

for sentence in batch['input_ids']:
    print(tokenizer.decode(sentence.tolist()))
    print('-'*100)
```
